# Exploratory Data Analysis

## About dataset


A real online retail transaction data set of two years.
Dataset Characteristics

Integer, Real

**Instances:** 1067371

Dataset Information

This Online Retail II data set contains all the transactions occurring for a UK-based and registered, non-store online retail between 01/12/2009 and 09/12/2011. The company mainly sells unique all-occasion gift-ware. Many customers of the company are wholesalers.

**Variable Information**

**InvoiceNo:** Invoice number. Nominal. A 6-digit integral number uniquely assigned to each transaction. If this code starts with the letter 'c', it indicates a cancellation.


**StockCode:** Product (item) code. Nominal. A 5-digit integral number uniquely assigned to each distinct product. 


**Description:** Product (item) name. Nominal. 


**Quantity:** The quantities of each product (item) per transaction. Numeric.	


**InvoiceDate:** Invoice date and time. Numeric. The day and time when a transaction was generated. 


**UnitPrice:** Unit price. Numeric. Product price per unit in sterling (£). 


**CustomerID:** Customer number. Nominal. A 5-digit integral number uniquely assigned to each customer. 


**Country:** Country name. Nominal. The name of the country where a customer resides.

# **Observations**

Each invoice can have more than one product, but each row corresponds to a unique product inside an invoice. That is why we have 1067371 rows but 53628 unique Invoice IDs. We need to identify the unique Customer IDs associated with each invoice in order to know the customer base. We need to drop rows where Customer IDs are NaNs, these are 243007 rows, accounting for 243007/1067371 = 22.7% rows in the dataset.

Date distributions seem to be rather uniform for the first and third quartiles: there are eight months of data in the first and third quartiles. For the second and fourth quartiles we count with five months, which suggest that transactions are more concentrated in the months of July-December.

I suspect that rows with negative quantities are those that have been canceled only. This might be a design decision but these quantities are not reliable for sale prediction since they have been canceled. However, I think that they might be valuable since there are 22950 items (rows) with these characteristics. This might be useful if we want to predict whether an order will be canceled or not. 22950 / ~1M rows gives around 2.295% canceled items (not invoices, though). Also, this might be helpful to identify customers that cancel orders.

There is one unique case where an Invoice ID starts with "C" and the quantity is positive, this might be an error. There are 19493 rows where the Invoice ID starts with "C" and the Quantity is negative. There are 22950 where the Quantity is negative. We have to discard all of those cases (starts with "C" OR Quantity is negative (nonpositive since there are no 0s)) since we can't rely on those cases.

We also have to drop cases where price is negative since these cases account for adjusting for bad debt.

 IMPORTANT: There are duplicated item registries inside invoices. For example, Invoice ID 489517 counts with items registered twice or three times. I don't know if the customer actually bought the same item three times in the same purchase and it was registered as three separate rows of quantity 1 (where it should have been registered as one row with quantity 3) or if there was a triplication error and the customer actually bought one of that product. The final decision regarding this is that we will group duplicated product rows in an invoice by quantity (sum their quantities and store them in one row) since we can't determine the real case scenario.

We have to check for NaNs all across the dataset. We initially count with 22.7% NaN rows that will be deleted because Customers are not identified. The other variable that contains NaN values is Description, which might not be a problem for what we are trying to solve at the moment.

There are 6202 rows where Price is 0. We have to check what these cases account for but it is likely that they will be dropped. 71 out of those cases actually have a Customer associated with them, but these are not valuable since they might have been gifts or something. The resting 6131 will be discarded anyways. The 6202 will be discarded.

Country variable does not present any problems, there are 43 unique values. There are 756 rows that are Unspecified. We could group countries with less volume under an "Other" category so we have less countries and handle them with one-hot encoding. This could be done after the initial data cleansing phase where we get to work with the customer base.

Stock Code variable is extremely heterogeneous. Code structures: codes that have 5 digits are standard, they comprise the majority of products (roughly 90%). Codes that have 6 digits (roughly 10%) are 5 digit codes that have a letter at the end which represent special products. Codes with length 4 have special cases like postage or gifts (the latter being discarded). Lengths 1, 3, 8, 9 are also special cases, handled by this stage of data cleansing. Length 7 presents interesting codes: on one hand we have regular items, but codes 'TEST001', 'TEST002' and 'ADJUST2' should be discarded. These are cases related to database testing or adjustments. Length 12 contains gifts (handled) and 'BANK CHARGES', the latter will be discarded since they are not sales. 

We could check if Stock Codes and Descriptions match but the latter won't be used for prediction most likely. We decide not to check this.

Variable types: Invoice, StockCode, Description, CustomerID and Country should be strings, Quantity should be an integer, InvoiceDate should be a datetime object and Price should be a float.
